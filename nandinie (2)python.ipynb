```python
# Common dict method

#1) keys : Returns a new object contraining all the keys in the dict
person = {  "name" : "alice" , "age" : 30 , "city" : "new york" }
print( person.keys()  )
```

## Finding Uppercase and Lowercase Characters in a String

```python
# Take a word input from the user and check how many are uppercase and how mnay are lower

# example : ElePhant    u = 2  l = 6


text = input("Enter a string ")


upper = 0
lower = 0

for i in text:
  if i.isupper():   # is upper
    upper = upper + 1
  elif i.islower():   # is lower
    lower = lower + 1


print("Uppercase letters", upper )
print("lowercase letters", lower )
```

map() function

the map() function applies a given functon to each item of an iterable ( like a list or a tuple) and returns a map object ( which can be converted into a list or other iterable types)

```python
def square(x):
  return x * x


nums = [1,2,3,4,5]

squarred_numbers = map( square , nums )

print(list(squarred_numbers))        # list is compulsary here
```

```python
class Student:
  def __init__(self,Name,Age,Hobby):    # constructer = _init_
    self.Name = Name  # self -> s3                              s3.Name = Name
    self.Age = Age                  # self - pointer or variable
    self.Hobby = Hobby

s1 = Student("a",20,"dancing")   # s1 is object
print(s1.Name)
print(s1.Age)
print(s1.Hobby)

print("****************")


s2 = Student("b"  , 25 , "Singing" )
print(s2.Name )
print(s2.Age )
print(s2.Hobby )

print("****************")


s3 = Student("c"  , 22 , "Cricket" )
print(s3.Name )
print(s3.Age )
print(s3.Hobby )
```

```python
# method in class

# function inside a class is called method
# normal function which is not called inside of anything

class Student:
  def __init__(self,Name,Age,Hobby,Marks): # constructur
    self.Name = Name                       # automatically called
    self.Age  = Age
    self.Hobby = Hobby        # we should have only one constructor in class
    self.Marks = Marks
  def avgMarks(self):    # method
    sum = 0
    for item in self.Marks:   # we need to call this
      sum = sum + item
    avg = sum / len(self.Marks)
    return avg


s1 = Student("A" , 21 , "Dancing" , [70,80,90,85] )
print(s1.Name)
print(s1.Age)
print(s1.Hobby)  # constructor = properties are added here
print(s1.Marks)
print( s1.avgMarks() )
```

Encapsulation

Encapsulation involves restricting direct access to an object's data and methods . you can make attributes private by prefixing them with a double underscore(__)

```python
class Bank():
  def __init__(self,name,age,balance):  # constructor
    self.name = name
    self.age = age
    self.__balance = balance    # private attribute / data by using '_'
  # Getter method to access the private attributes
  def get_balance(self):
    return self.__balance
  def add_balance(self,amount):             # function inside a class
    self.__balance = self.__balance + amount
  def withdraw(self,amount):
    self.__balance = self.__balance - amount

  # self is a parameter and balance is a variable


acc1 = Bank( "a" , 30 , 1000 )
print( acc1.name  )
print( acc1.age  )
print( acc1.get_balance()  )
acc1.add_balance(500)
print( acc1.get_balance()  )
acc1.withdraw(200)
print( acc1.get_balance()  )
```

Polymorphism

Polymorphism in python simply means "many forms" . it allows different objects to use the same method or function in different ways

Lets break it down in the simplest way: you have a cat and dog. both are animals,but tthey make different sounds based on what object we pass.

```python
class Dog:
  def sound(self):
    return "Bark"

class Cat :
  def sound(self):
    return "Meow"

# function that take an animal and asks for the sound
def make_animal_sound(animal): # animal = cat
  print( animal.sound() ) # animal.sound ->    cat.sound()

# Create objects of Dog and Cat
dog = Dog()
cat = Cat()


# call the same function , but it works for both animals
make_animal_sound( cat )
make_animal_sound( dog)
```
